Exercise 2: Calculate the Area of a Circle
Ask the user to enter the radius of a circle.

Use:

math.pi
to calculate:

area = π × radius²
Display the area to two decimal places.

Example:

Enter the radius: 5
Area: 78.54

```python
import math

radius = int(input("Enter the radius: "))
area = math.pi * math.pow(radius,2)
print(f"Area: {round(area,2)}")
```

Exercise 6: Import Specific Functions
Instead of importing the entire math module, use:

In a text cell, briefly explain one difference between:

import math

```python
from math import sqrt

print(f"Square Root of 25: {sqrt(25)}")
print(f"Square Root of 100: {sqrt(100)}")
print(f"Square Root of 144: {sqrt(144)}")
```

Exercise 7: Build a Random Number Utility
Create a small program that asks the user for:

A minimum number
A maximum number
Use random.randint() to generate a random number within that range.

Example:

Enter the minimum number: 10
Enter the maximum number: 50

Random Number: 37
Test the program with at least two different ranges.



```python
import random

min_number = int(input("Enter the minimum number: "))
max_number = int(input("Enter the maximum number: "))

random_number = random.randint(min_number,max_number)
print(f"Random Number: {random_number}")
```
