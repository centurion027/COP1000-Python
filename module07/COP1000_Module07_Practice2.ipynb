Exercise 3: Remove Duplicate Values
Create the following list:

names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]
Display the original list.

Convert the list to a set.

Display the set.

Then display:

The number of values in the original list
The number of unique values in the set
Observe what happens to duplicate values.

```python
names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]

print(names)

org_list_count = len(names)

names = set(names)

print(names)

print(org_list_count)
print(len(names))
```

Exercise 4: Test Membership
Create:

languages = {"Python", "Java", "C++", "JavaScript"}
Ask the user to enter the name of a programming language.

Use the in operator to determine whether the language appears in the set.

Example:

Enter a programming language: Python
Python is in the set.
or:

Enter a programming language: Ruby
Ruby is not in the set.

```python
languages = {"Python", "Java", "C++", "JavaScript"}

guess = input("Enter a programming language: ")

if guess in languages:
  print(f"{guess} is in the set.")
else:
  print(f"{guess} is not in the set.")
```

Exercise 7: Compare Course Enrollments
Create:

python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}
Use set operations to display:

Students enrolled in both courses
Students enrolled in either course
Students who are in Python but not Web
Add descriptive labels to each result.

Example:

Students in both courses: ...
Students in either course: ...
Python only: ...

```python
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}

first = python_students.intersection(web_students)
second = python_students.union(web_students)
third = python_students.difference(web_students)

print(f"Students in both course: {first}")
print(f"Students in either course: {second}")
print(f"Python only: {third}")
```
