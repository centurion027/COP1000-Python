Exercise 6: Loop Through a Dictionary
Create:

prices = {
    "Notebook": 3.49,
    "Pen": 1.25,
    "Folder": 2.00,
    "Backpack": 24.99
}
Use a loop to display each item and its price.

Example output:

Notebook: $3.49
Pen: $1.25
Folder: $2.00
Backpack: $24.99
Format each price to two decimal places.

```python
prices = {
    "Notebook": 3.49,
    "Pen": 1.25,
    "Folder": 2.00,
    "Backpack": 24.99
}

for item, price in prices.items():
  print(f"{item} ${price:.2f}")
```

Exercise 7: Build a Simple Contact Record
Create an empty dictionary named:

contact = {}
Ask the user to enter:

Name
Phone number
Email address
City
Store each response in the dictionary using an appropriate key.

Then display the completed contact information in a clean format.

Example:

Contact Information

Name: Jordan Smith
Phone: 555-1234
Email: jordan@example.com
City: Jacksonville
Your program must use dictionary values in the final output.

```python
contact = {}

contact["name"] = input("Enter Name: ")
contact["phone_num"] = input("Enter Phone Number: ")
contact["email"] = input("Enter Email: ")
contact["city"] = input("Enter City: ")

print("\nContact Information\n")

print(f"Name: {contact["name"]}")
print(f"Phone: {contact["phone_num"]}")
print(f"Email: {contact["email"]}")
print(f"City: {contact["city"]}")
```
