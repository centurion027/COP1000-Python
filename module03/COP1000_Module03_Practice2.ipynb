Exercise 2: if and else
Ask the user to enter their age.

Convert the value to an integer.

Write a program that displays:

You are eligible.
if the user is 18 or older.

Otherwise, display:

You are not eligible.
Test the program with:

A value greater than 18
A value less than 18

```python
age = int(input("Age: "))

if age >= 18:
  print("You are eligible")
else:
  print("You are not eligible.")
```

Exercise 3: Use elif
Ask the user to enter a numeric grade from 0 through 100.

Use conditional statements to display:

A for 90 or higher
B for 80 through 89
C for 70 through 79
D for 60 through 69
F for below 60
Example:

Enter your grade: 84
Letter Grade: B
Test your program with at least three different grades.

```python
grade = int(input("Enter your grade: "))

if grade >= 90:
  print("Letter Grade: A")

elif grade >= 80:
  print("Letter Grade: B")

elif grade >= 70:
  print("Letter Grade: C")

elif grade >= 60:
  print("Letter Grade: D")

elif grade >= 0:
  print("Letter Grade: F")
```

Exercise 4: Check a Number
Ask the user to enter a whole number.

Write a conditional statement that determines whether the number is:

Positive
Negative
Zero
Example:

Enter a number: -5
The number is negative.
Test all three possibilities.

```python
number = int(input("Enter a number: "))

if number > 0:
  print("The number is positive.")

elif number < 0:
  print("The number is negative.")

elif number == 0:
  print("The number is zero.")
```

Exercise 5: Use and
Ask the user for:

Their age
Whether they have a valid ID
For the ID question, have the user enter yes or no.

Display:

Entry permitted.
only if:

the user is at least 18 years old, and
the user entered yes
Otherwise display:

Entry denied.

```python
age = int(input("Age: "))
id = input("Do you have a valid ID: ")

age_req = age >= 18
id_req = id == "yes"

if age_req and id_req:
  print("Entry permitted.")

else:
  print("Entry denied.")
```

Exercise 6: Make a Weather Recommendation
Ask the user to enter the current temperature.

Use conditional statements to display:

Wear a coat. if the temperature is below 50.
A light jacket may be useful. if the temperature is 50 through 69.
The weather is comfortable. if the temperature is 70 through 84.
It is hot outside. if the temperature is 85 or higher.
Test the program with at least three different temperatures.

```python
temp = int(input("What is the current temperature: "))

if temp < 50:
  print("Wear a coat.")
elif temp < 70:
  print("A light jacket may be useful.")
elif temp < 85:
  print("The weather is comfortable.")
elif temp >= 85:
  print("It is hot outside.")
```
