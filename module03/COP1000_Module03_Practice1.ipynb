Exercise 2: Comparison Operators
Create:

age = 21
Write and display the result of comparisons that answer the following questions:

Is age equal to 21?

Is age not equal to 18?

Is age greater than 18?

Is age less than 30?

Is age greater than or equal to 21?

Is age less than or equal to 20?

Each expression should produce either:

True
or:

False

```python
age = 21

print(f"Is {age} equal to 21?")
print(age == 21)

print(f"Is {age} not equal to 18?")
print(age != 18)

print(f"Is {age} greater than 18?")
print(age > 18)

print(f"Is {age} less than 30?")
print(age < 30)

print(f"Is {age} greater than or equal to 21?")
print(age >= 21)

print(f"Is {age} less than or equal to 20?")
print(age <= 20)
```

Exercise 4: Boolean Operators
Create the following variables:

age = 22
has_id = True
Use Python to evaluate:

age >= 18 and has_id
Then evaluate:

age < 18 or has_id
Then evaluate:

not has_id
Display each result.

Add a comment identifying which Boolean operator is being demonstrated:

and

or

not

```python
age = 22
has_id = True

# This is an "and" Boolean operator
print(age >= 18 and has_id)

# This is an "or" Boolean operator
print(age < 18 or has_id)

# This is a "not" Boolean operator
print(not has_id)
```

Exercise 5: Combine Comparisons
Create:

temperature = 75
Write Boolean expressions that determine whether:

The temperature is greater than 70.

The temperature is less than 90.

The temperature is between 70 and 90.

The temperature is outside the range 60 through 85.

Display the result of each expression.

```python
temperature = 75

# Greater than 70
print(temperature > 70)

# Less than 90
print(temperature < 90)

# Between 70 and 90
print(temperature > 70 and temperature < 90)

# Outside the range 60 through 85
print(temperature < 60 or temperature > 85)
```

Exercise 6: Build a Simple Eligibility Check
Create:

age = 19
credits_completed = 30
Write Boolean expressions to determine whether the student:

Is at least 18 years old.

Has completed at least 24 credits.

Meets both requirements.

Display each result with a descriptive label.

Example output:

Age Requirement Met: True
Credit Requirement Met: True
All Requirements Met: True

```python
age = 19
credits_completed = 30

age_requirement = age >= 18
credit_requirement = credits_completed >= 24

all_requirements = age_requirement and credit_requirement

print(f"Age Requirement Met: {age_requirement}")
print(f"Credit Requirement Met: {credit_requirement}")
print(f"All Requirements Met: {all_requirements}")
```
