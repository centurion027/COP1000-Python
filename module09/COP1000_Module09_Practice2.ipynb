Exercise 5: Debug a Conditional Statement
The following program should determine whether a student passed:

grade = 82

if grade <= 60:
    print("Passed")
else:
    print("Failed")
Correct the logic so that a grade of 82 produces:

Passed
Test the program with:

82
60
59

```python
def grade_checker(grade):
  if grade >= 60:
    print("Passed")
  else:
    print("Failed")

test_list = [82, 60, 59]

for grade in test_list:
  grade_checker(grade)
```

Exercise 7: Test a Program With Multiple Values
Create a function:

def calculate_discount(price, discount_percent):
The function should calculate and return the price after the discount.

Test the function using these cases:

Price	Discount	Expected Result
$100	20%	$80.00
$50	10%	$45.00
$25	0%	$25.00
Display the actual result for each test.

Then answer in a text cell:

Did each test produce the expected result?

```python
def calculate_discount(price, discount_percent):
  discount = price * discount_percent
  total = price - discount
  return total

test_list = [(100, .20), (50, .10), (25, 0)]

for price, discount_percent in test_list:
  total = calculate_discount(price, discount_percent)
  print(f"Price after discount: ${total:.2f}")
```
