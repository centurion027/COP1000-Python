Exercise 3: Protect a Division Calculation
Ask the user to enter two numbers.

Divide the first number by the second.

Use try and except to handle:

Invalid numeric input
Division by zero
Your program should display an appropriate message instead of crashing.

Example:

Enter the first number: 10
Enter the second number: 0

You cannot divide by zero.

```python
try:
  num1 = int(input("Enter the first number: "))
  num2 = int(input("Enter the second number: "))
  answer = num1/num2
  print(f"{num1} divided by {num2} = {answer}")

except ValueError:
  print("\nYou must enter a number.")

except ZeroDivisionError:
  print("\nYou cannot divide by zero.")
```

Exercise 4: Validate a Grade
Ask the user to enter a numeric grade from 0 through 100.

Your program must:

Use try and except to handle nonnumeric input.
Check whether the numeric value is within the valid range.
Display a message indicating whether the grade is valid.
Example:

Enter a grade: 105
Grade must be between 0 and 100.
Test:

A valid grade
A value above 100
A negative value
Nonnumeric input

```python
try:
  grade = int(input("Enter a numeric grade from 0 through 100: "))
  if 0 <= grade <= 100:
    print("Grade is valid.")
  else:
    print("Grade must be between 0 and 100.")
except ValueError:
  print("You must enter a number.")
```

Exercise 5: Keep Asking Until Input Is Valid
Create a program that repeatedly asks the user for their age.

Use a loop together with try and except.

The program should continue until the user enters a valid positive whole number.

Example:

Enter your age: twenty
Invalid input. Enter a whole number.

Enter your age: -5
Age must be greater than zero.

Enter your age: 21
Age accepted: 21

```python
while True:
  try:
    age = int(input("Enter your age: "))
    if age > 0:
      print(f"Age accepted: {age}")
      break
    else:
      print("Age must be greater than zero.")
  except ValueError:
    print("Invalid input. Enter a whole number.")
```

Exercise 6: Handle a Missing File
Write a program that attempts to open:

grades.txt

for reading.

Use:

except FileNotFoundError:
to handle the situation if the file does not exist.

Display:

The requested file could not be found.
instead of allowing the program to stop unexpectedly.

```python
try:
  with open("grades.txt", "r") as file:
    print(file.read())
except FileNotFoundError:
  print("The requested file could not be found.")
```

Exercise 7: Safe Average Calculator
Ask the user for three test scores.

Use exception handling so that invalid numeric input does not crash the program.

If all three values are valid:

Calculate the average.
Display it to two decimal places.
Example:

Enter score 1: 80
Enter score 2: 92
Enter score 3: 85

Average: 85.67
If invalid input is entered, display a clear error message.

```python
scores = 0

for i in range(3):
  while True:
    try:
      score = int(input(f"Enter score {i + 1}: "))
      scores += score
      break
    except ValueError:
      print("Please enter a number.")
avg_score = scores/(3)

print(f"Average : {avg_score:.2f}")
```
