Exercise 6: Work With Numbers in a List
Create:

scores = [88, 94, 76, 91, 85]
Use Python to display:

The highest score using max()
The lowest score using min()
The total using sum()
The number of scores using len()
Then calculate the average:

average = total / number of scores
Display the average to two decimal places.

```python
scores = [88, 94, 76, 91, 85]

max_score = max(scores)
low = min(scores)
total = sum(scores)
count = len(scores)

print(max_score)
print(low)
print(total)
print(count)

average = total/count

print(f"{average:.2f}")
```

Exercise 7: Build a Simple Student List
Create an empty list:

students = []
Ask the user to enter three student names.

Add each name to the list using append().

After all three names have been entered:

Display the complete list
Display the number of students
Sort the names alphabetically
Display the sorted list

```python
students = []

while True:
  if len(students) <3:
    new_name = input("Enter a name: ")
    students.append(new_name)
  else:
    print(students)
    print(len(students))

    students.sort()
    print(students)
    break
```
