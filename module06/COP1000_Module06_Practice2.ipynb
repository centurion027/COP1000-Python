Exercise 1: Loop Through a List
Create:

animals = ["dog", "cat", "rabbit", "horse", "turtle"]
Use a for loop to display each animal on a separate line.

Your loop should process the list rather than using separate print() statements.

```python
animals = ["dog", "cat", "rabbit", "horse", "turtle"]

for i in range(len(animals)):
  print(animals[i])
```

Exercise 2: Number the Items
Create:

tasks = ["Study Python", "Complete homework", "Read chapter", "Review notes"]
Use a loop to display the tasks with numbers.

Example:

1. Study Python
2. Complete homework
3. Read chapter
4. Review notes
You may use range() or enumerate().

```python
tasks = ["Study Python", "Complete homework", "Read chapter", "Review notes"]

for i, idx in enumerate(tasks):
  i += 1
  print(i, idx)
```

Exercise 3: Search a List
Create:

Exercise 4: Filter Values
Create:

temperatures = [45, 72, 88, 61, 93, 55, 79]
Use a for loop and conditional statement to display only temperatures that are 70 or higher.

Then count how many temperatures met the requirement.

Display the count after the loop.

```python
temperatures = [45, 72, 88, 61, 93, 55, 79]
count = 0
for i, idx in enumerate(temperatures):
  i += 1
  if idx >= 70:
    count += 1
    print(idx)
print(count)
```

Exercise 5: Calculate an Average From User Data
Create an empty list named:

scores = []
Ask the user to enter five test scores.

Add each score to the list.

After all five scores have been entered:

Display the list
Calculate the total
Calculate the average
Display the highest score
Display the lowest score
Format the average to two decimal places.

```python
scores = []
score_count = 5

for i in range(score_count):
  new_score = int(input("Enter test score:"))
  scores.append(new_score)

score_total = sum(scores)
score_avg = score_total / score_count
score_min = min(scores)
score_max = max(scores)

print(scores)
print(score_total)
print(f"{score_avg:.2f}")
print(score_max)
print(score_min)
```

Exercise 6: Create a Tuple
Create a tuple containing the days of the weekend:

weekend = ("Saturday", "Sunday")
Display:

The complete tuple
The first item
The second item
The number of items using len()
Then use a for loop to display each day.

```python
weekend = ("Saturday", "Sunday")

print(weekend)
print(weekend[0])
print(weekend[1])
print(len(weekend))

for i in range(len(weekend)):
  print(weekend[i])
```
