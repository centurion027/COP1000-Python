Exercise 6: Calculate a Running Total
Create the following variable:

total = 0
Use a for loop to add the numbers from 1 through 10 to total.

After the loop finishes, display the final result.

Your output should clearly identify the answer:

The total is 55.
Do not manually assign 55 to the variable.

```python
total = 0

for i in range(1, 11):
  total += i

print(f"The total is {total}")
```

Exercise 7: Calculate Five Test Scores
Create a variable named:

total = 0
Use a for loop that repeats five times.

During each repetition:

Ask the user to enter a test score.
Convert the score to a float.
Add the score to total.
After all five scores have been entered:

Calculate the average.
Display the total.
Display the average formatted to two decimal places.
Example:

Enter test score 1: 80
Enter test score 2: 90
Enter test score 3: 75
Enter test score 4: 95
Enter test score 5: 85

Total: 425.00
Average: 85.00

```python
total = 0

for i in range(1, 6):
  score = int(input(f"Enter test score {i} "))
  score = float(score)
  total = score + total
print(f"Total: {total:.2f}")
print(f"Average: {total / 5:.2f}")
```
