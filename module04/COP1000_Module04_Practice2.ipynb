Exercise 1: Count With a while Loop
Create:

number = 1
Use a while loop to display the numbers 1 through 10.

Remember that the value of number must change inside the loop.

```python
number = 1

while number <= 10:
  print(number)
  number += 1
```

Exercise 2: Countdown
Create:

count = 5
Use a while loop to count backward from 5 to 1.

After the loop finishes, display:

Done!

```python
count = 5

while count >= 1:
  print(count)
  count -= 1

print("Done!")
```

Exercise 3: Repeat Until the User Stops
Create a program that repeatedly asks:

Would you like to continue? Enter yes or no:
The program should continue while the user enters:

yes
When the user enters no, display:

Program finished.

```python
ex3 = input("Would you like to continue? Enter yes or no: ")

while ex3 != ("no"):

  ex3 = input("Would you like to continue? Enter yes or no: ")
print("Program finished")

```

Exercise 4: Password Check
Create:

password = "python123"
Ask the user to enter the password.

Use a while loop to continue asking until the correct password is entered.

Example:

Enter the password: hello
Incorrect password. Try again.

Enter the password: test
Incorrect password. Try again.

Enter the password: python123
Access granted.
Do not display the correct password to the user.

```python
passInput = input("Enter the password: ")
password = "python123"

while password != passInput:
  print("Invalid password. Try again.")
  passInput = input("Enter the password: ")

print("Access granted.")
```

Exercise 5: Validate Numeric Input
Ask the user to enter a number from 1 through 10.

If the number is outside that range, continue asking until the user enters a valid number.

Example:

Enter a number from 1 to 10: 15
Invalid number. Try again.

Enter a number from 1 to 10: -2
Invalid number. Try again.

Enter a number from 1 to 10: 7
You entered a valid number: 7
Use a while loop to perform the validation.

```python
numb = int(input("Enter a number from 1 to 10: "))

while numb not in range(1, 11):
  print("Invalid number. Try again.")
  numb = int(input("Enter a number from 1 to 10: "))

print(f"You entered a valid number: {numb}")
```

Exercise 6: Running Total Until Zero
Create:

total = 0
Ask the user to enter numbers one at a time.

Continue adding each number to the running total.

When the user enters 0, stop asking for numbers and display the total.

Example:

Enter a number or 0 to stop: 10
Enter a number or 0 to stop: 5
Enter a number or 0 to stop: 8
Enter a number or 0 to stop: 0

Total: 23
The value 0 should stop the loop.

```python
total = 0
numb = int(input("Enter a number or 0 to stop: "))

while numb != 0:
  total += numb
  numb = int(input("Enter a number or 0 to stop: "))

print(f"Total: {total}")
```

Exercise 7: Limited Attempts
Create a program that allows the user a maximum of three attempts to enter the word:

python
Use a counter to keep track of the number of attempts.

If the user enters the correct word, display:

Correct!
If all three attempts are used without a correct answer, display:

No attempts remaining.
Test both outcomes.

```python
key = "python"
userInput = input("Enter the word: ")
attempts = 3

while attempts > 0:
  if userInput == key:
    print("Correct!")
    break
  else:
    attempts -= 1

    if attempts == 0:
        print("No attempts remaining.")

    else:
        print(f"{attempts} Attempts remaining.")
        userInput = input("Enter the word: ")
```
