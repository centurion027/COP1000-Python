Exercise 5: Use Multiple Parameters
Create a function named:

def student_info(name, major):
The function should display the student's name and major using an f-string.

Example:

student_info("Taylor", "Cybersecurity")
could display:

Taylor is studying Cybersecurity.
Call the function at least twice using different values.

```python
def student_info(name, major):
  print(f"{name} is studying {major}.")

student_info("Taylor", "Cybersecurity")
student_info("Allen", "Accounting")
```

Exercise 6: Pass User Input to a Function
Ask the user to enter:

Their name
Their favorite programming topic so far
Store both responses in variables.

Create a function that accepts both values as parameters and displays a sentence using them.

Example:

Enter your name: Jordan
Enter your favorite topic: loops

Jordan's favorite Python topic so far is loops.
Your function should work with whatever values the user enters.

```python
user_name = input("Enter your name: ")
fav_topic = input("Enter your favorite topic: ")

def user_fav_topic(user_name, fav_topic):
  print(f"{user_name}'s favorite Python topic so far is {fav_topic}.")

user_fav_topic(user_name, fav_topic)
```

Exercise 7: Break a Program Into Functions
Create a small program using at least three functions:

One function to display a title
One function to display a short message
One function to display a closing message
Call the functions from the main part of your program.

Your final output should clearly show that the program has been divided into separate tasks.

```python
def dis_title():
  print("This is the title")

def sht_msg():
  print("This is the short message")

def cls_msg():
  print("This is the closing message")

def msg_pgm():
  dis_title()
  sht_msg()
  cls_msg()

msg_pgm()
```
