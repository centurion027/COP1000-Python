Exercise 1: Return a Value
Create a function named:

def double_number(number):

Exercise 2: Add Two Numbers
Create a function named:

def add_numbers(number1, number2):

Ask the user to enter two numbers.

Example:

Exercise 4: Convert Fahrenheit to Celsius
Create a function named:

def fahrenheit_to_celsius(fahrenheit):
Use the following formula:

Celsius = (Fahrenheit - 32) × 5 / 9
Return the Celsius temperature.

Ask the user for a Fahrenheit temperature and display the converted result to one decimal place.

Example:

Enter temperature in Fahrenheit: 72

Temperature in Celsius: 22.2

```python
def fahrenheit_to_celsius(fahrenheit):
  celsius = (fahrenheit - 32) * 5 / 9
  return celsius

fahrenheit = int(input("Enter temperature in Fahrenheit: "))

print(f"Temperature in Celsius : {fahrenheit_to_celsius(fahrenheit):.1f}")
```

Exercise 5: Determine a Letter Grade
Create a function named:

def get_letter_grade(score):
The function should return:

"A" for 90 or higher
"B" for 80 through 89
"C" for 70 through 79
"D" for 60 through 69
"F" for below 60
Ask the user for a numeric grade.

Pass the grade to the function and display the returned letter grade.

Test the function with at least three different scores.

```python
def get_letter_grade(score):
  if score >= 90:
    return "A"
  if score >= 80:
    return "B"
  if score >= 70:
    return "C"
  if score >= 60:
    return "D"
  else:
    return"F"

score = int(input("Enter the score: "))

print(f"The score is a letter grade of {get_letter_grade(score)}")
```

Exercise 6: Separate Calculation From Output
Create two functions:

def calculate_total(price, quantity):
and:

def display_total(item, total):
The first function should calculate and return the total cost.

The second function should display the item name and total.

Ask the user for:

Item name
Price
Quantity
Your program should use both functions.

Example output:

Enter the item: Notebook
Enter the price: 3.50
Enter the quantity: 4

Item: Notebook
Total Cost: $14.00

```python
def calculate_total(price, quantity):
  total = price * quantity
  return total

def display_total(item, total):
  print(f"Item: {item}")
  print(f"Total Cost: ${total:.2f}")

item = input("Enter the item: ")
price = float(input("Enter the price: "))
quantity = int(input("Enter the quantity: "))
total = calculate_total(price, quantity)

display_total(item, total)
```

A function that displays a title.
A function that performs a calculation and returns a value.
A function that displays the final result.
You may choose the calculation.

Examples include:

Miles to kilometers
Hours to minutes
Price after discount
Total cost of several items
Age in months

```python
def title_display():
  print("Hours to minutes calculator")

def hr_to_min_calc(hrs):
  mins = hrs * 60
  return mins

def answer_display(mins):
  print(f"{hrs} hour's equals {mins} minutes's")

hrs = int(input("Enter number of hours: "))
mins = hr_to_min_calc(hrs)

title_display()
answer_display(mins)
```
