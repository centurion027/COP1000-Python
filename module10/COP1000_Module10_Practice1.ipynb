Exercise 4: Add More Attributes
Modify the Book class so that each object also stores:

Publication year

Use appropriate data types for each attribute.

Display the information in a clean format.

Exercise 5: Update an Object Attribute
Create a class named:

class Product:
Each product should store:

Name
Price
Quantity
Create one product object.

Display its original quantity.

Then change the quantity and display the updated value.

Example:

Original Quantity: 10
Updated Quantity: 7

```python
class Product:
  def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

rice = Product("Rice", 2.99, 3)

print(f"Original Quantity: {rice.quantity}")

rice.quantity = 2

print(f"Updated Quantity: {rice.quantity}")
```

```python
book1 = Book("Harry Potter", "J. K. Rowling", "1997", True)
book2 = Book("1984", "George Orwell", "1949", True)
book3 = Book("The Hobbit", "J. R. R. Tolkien", "1937", False)

books = [book1, book2, book3]

for book in books:
  print(book.title, book.author)
```
