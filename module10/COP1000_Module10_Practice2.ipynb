Exercise 2: Create a Calculation Method
Create a class named:

class Rectangle:
The class should store:

Length
Width
Create a method named:

def calculate_area(self):
The method should return the rectangle's area.

Create at least two rectangle objects and display their areas.

```python
class Rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def calculate_area(self):
    area = self.length * self.width
    return area

rectangle1 = Rectangle(8, 2)
rectangle2 = Rectangle(3, 7)

print(rectangle1.calculate_area())
print(rectangle2.calculate_area())
```

Exercise 3: Create a Method That Changes an Attribute
Create a class named:

class BankAccount:
Store:

Account holder
Balance
Create a method:

def deposit(self, amount):
The method should add the deposit amount to the balance.

Create an account with a starting balance.

Make a deposit.

Display the updated balance.

Example:

Starting Balance: $500.00
Deposit: $100.00
New Balance: $600.00

```python
class BankAccount:
  def __init__(self, acctholder, balance):
    self.acctholder = acctholder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f"Deposit: ${amount:.2f}")
    print(f"New Balance: ${self.balance:.2f}")
    return self.balance

account1 = BankAccount("Steven", 500)

print(f"Starting Balance: ${account1.balance:.2f}")

account1.deposit(100)
```

Use a conditional statement so that a withdrawal cannot exceed the available balance.

Insufficient funds.
Test both:

A successful withdrawal
An unsuccessful withdrawal

```python
class BankAccount:
  def __init__(self, acctholder, balance):
    self.acctholder = acctholder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f"Deposit: ${amount:.2f}")
    print(f"New Balance: ${self.balance:.2f}")
    return self.balance

  def withdraw(self, amount):
    if self.balance < amount:
      print("Insufficient funds.")
    else:
      self.balance -= amount
      print(f"Withdrawal: ${amount:.2f}")
      print(f"New Balance: ${self.balance:.2f}")

account1 = BankAccount("Steven", 500)

print(f"Starting Balance: ${account1.balance:.2f}")

#account1.deposit(100)
account1.withdraw(501)
```

Exercise 5: Model a Book
Create a class named:

class Book:
Store:

Title
Author
Availability
Add methods:

def check_out(self):
and:

def return_book(self):
When check_out() is called, the book should become unavailable.

When return_book() is called, the book should become available again.

Display the book's status after each action.

```python
class Book:
  def __init__(self, title, author, availability):
    self.title = title
    self.author = author
    self.availability = availability

  def check_out(self):
    self.availability = False
    print(f"{self.title} is unavailable.")

  def return_book(self):
    self.availability = True
    print(f"{self.title} is available.")

book1 = Book("Harry Potter", "J. K. Rowling", True)

book1.check_out()
book1.return_book()
```

Exercise 6: Multiple Objects With Methods
Create three Book objects.

Place them in a list.

Use a loop to call a method that displays the title, author, and availability of every book.

Then check out one book and display all three books again.

The status of the checked-out book should change.

```python
class Book:
  def __init__(self, title, author, availability):
    self.title = title
    self.author = author
    self.availability = availability

  def check_out(self):
    self.availability = False
    print(f"{self.title} is unavailable.\n")

  def return_book(self):
    self.availability = True
    print(f"{self.title} is available.\n")

  def book_display(self):
    print(f"Title: {self.title}\nAuthor: {self.author}\nAvailable: {self.availability}\n")

book1 = Book("Harry Potter", "J. K. Rowling", True)
book2 = Book("1984", "George Orwell", True)
book3 = Book("The Hobbit", "J. R. R. Tolkien", True)

books = [book1, book2, book3]

for book in books:
  book.book_display()

book2.check_out()

for book in books:
  book.book_display()
```

Exercise 7: Design Your Own Class
Create a class representing one of the following:

Employee
Vehicle
Game Character
Fitness Tracker
Shopping Cart Item
Your class must have:

At least three attributes
An __init__ method
At least two methods
At least one method that changes an attribute
Create an object and demonstrate both methods.

```python
class Employee:
  def __init__(self, name, position, promoted):
    self.name = name
    self.position = position
    self.promoted = promoted

  def promotion(self):
    self.promoted = True
    print(f"{self.name} has been promoted to {self.position}.\n")

  def promotion_roster(self):
    if self.promoted == True:
      print(f"Name: {self.name}\nPosition: {self.position}\n")


employee1 = Employee("John Smith", "Manager", False)
employee2 = Employee("Sarah Johnson", "Supervisor", False)
employee3 = Employee("Michael Brown", "Analyst", False)

employees = [employee1, employee2, employee3]

for employee in employees:
  employee.promotion_roster()

employee2.promotion()

for employee in employees:
  employee.promotion_roster()
```
