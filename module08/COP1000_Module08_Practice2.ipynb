Exercise 2: Create a CSV File
Create a file named:

students.csv

Write the following rows:

Name,Major,Credits
Alex,Cybersecurity,24
Jordan,Information Technology,36
Taylor,Computer Science,42
Use Python's csv.writer() rather than manually creating the entire file as one string.

```python
with open("students.csv", "w") as file:
  students = csv.writer(file)
  students.writerow(["Name","Major","Credits"])
  students.writerow(["Alex","Cybersecurity",24])
  students.writerow(["Jordan","Information Technology",36])
  students.writerow(["Taylor","Computer Science",42])
```

Exercise 3: Read a CSV File
Open students.csv in read mode.

Use:

csv.reader()
to display every row.

Your output should clearly show the header and all three student records.

```python
with open("students.csv", "r") as file:
  students = csv.reader(file)
  for student in students:
    print(student)
```

Exercise 4: Display Selected Data
Read students.csv again.

Skip the header row and display only:

Student name

Exercise 5: Add a New Record
Ask the user to enter:

Student name
Major
Number of completed credits
Append the new record to students.csv.

Then read and display the updated CSV file.

```python
with open("students.csv", "a") as file:
  students = csv.writer(file)
  name = input("Enter name: ")
  major = input("Enter major: ")
  credits = int(input("Enter credits: "))

  students.writerow([name, major, credits])

with open("students.csv", "r") as file:
  students = csv.reader(file)
  for student in students:
    print(student)
```

Exercise 6: Calculate From CSV Data
Read the credit values from students.csv.

Convert each credit value to an integer.

Calculate and display:

Total credits represented in the file
Number of student records
Average number of credits
Format the average to two decimal places.

Example:

Student Records: 4
Total Credits: 132
Average Credits: 33.00
Your exact results will depend on the record you added.

```python
with open("students.csv", "r") as file:
    students = csv.reader(file)
    next(students)

    total_credits = 0
    student_count = 0

    for student in students:
        total_credits += int(student[2])
        student_count += 1

average_credits = total_credits / student_count

print(f"Student Records: {student_count}")
print(f"Total Credits: {total_credits}")
print(f"Average Credits: {average_credits:.2f}")
```

Exercise 7: Create Your Own CSV Dataset
Create a new CSV file named:

products.csv

The file must include:

A header row
At least three product records
Product name
Price
Quantity
Example structure:

Product,Price,Quantity
Notebook,3.49,4
Pen,1.25,10
Folder,2.00,5
After creating the file, read it with Python and display each product record.

```python
import csv

with open("products.csv", "w", newline="") as file:
    products = csv.writer(file)
    products.writerow(["Product", "Price", "Quantity"])
    products.writerow(["Notebook", 3.49, 4])
    products.writerow(["Pen", 1.25, 10])
    products.writerow(["Folder", 2.00, 5])

with open("products.csv", "r") as file:
    products = csv.reader(file)
    for product in products:
        print(product)
```
