Exercise 1: Work With Strings
Create:

message = "Python programming is useful"

The entire string

useful

powerful

Exercise 6: Append to a File
Open student.txt using append mode:

"a"
Add one additional line to the file.

For example:

Module 8: File I/O
Then reopen the file in read mode and display the complete updated contents.

```python
with open("student.txt", "w") as file:
  file.write("Steven\n")
  file.write("COP 1000\n")
  file.write("Is learning a lot of new material in this class")

print("File created successfully.")

with open("student.txt", "a") as file:
  file.write("\nModule 8: File I/O")

with open("student.txt", "r") as file:
  contents = file.read()

print(contents)
```

Exercise 7: Create a Simple Notes File
Ask the user to enter three short notes.

Write all three notes to a file named:

notes.txt

Each note should appear on a separate line.

Then read the file and display all three notes.

Example:

Enter note 1: Review Python
Enter note 2: Finish assignment
Enter note 3: Study file handling

Saved Notes:
Review Python
Finish assignment
Study file handling

```python
for i in range(3):
  if i == 0:
    with open("notes.txt" , "w") as file:
      note_input = input(f"Enter note {i + 1}: ")
      file.write("\nSaved Notes:\n")
      file.write(note_input + "\n")
  else:
    with open("notes.txt" , "a") as file:
      note_input = input(f"Enter note {i + 1}: ")
      file.write(note_input + "\n")

with open("notes.txt", "r") as file:
  contents = file.read()

print(contents)
```
