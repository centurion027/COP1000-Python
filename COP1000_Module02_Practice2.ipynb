Exercise 5: Calculate a Purchase
Ask the user to enter:

An item name
The item's price
The quantity being purchased
Convert the price to a float.

Convert the quantity to an int.

Calculate the total cost.

Example:

Enter the item: Notebook
Enter the price: 3.49
Enter the quantity: 4

Purchase Summary
Item: Notebook
Price: $3.49
Quantity: 4
Total Cost: $13.96
Format monetary values to two decimal places.

```python
item = input("Enter the item: ")
price = float(input("Enter the price: "))
quantity = int(input("Enter the quantity: "))

print("Purchase Summary")
print(f"Item: {item}")
print(f"Price: ${price:.2f}")
print(f"Quantity: {quantity}")
print(f"Total Cost: ${quantity * price:.2f}")
```

Exercise 6: Calculate Gross Pay
Ask the user to enter:

Number of hours worked
Hourly pay rate
Convert both responses to numeric values and calculate:

gross pay = hours worked × hourly rate

Display the results using an f-string.

Example:

Hours worked: 20
Hourly rate: 15.50

Gross Pay: $310.00
Python must perform the calculation.

```python
hrs = int(input("Hours worked: "))
rate = float(input("Hourly rate: "))

print(f"Gross Pay: ${hrs * rate:.2f}")
```
