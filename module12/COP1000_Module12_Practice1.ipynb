Exercise 2: Work With Nested Data
Create:

weather = {
    "city": "Jacksonville",
    "current": {
        "temperature": 82,
        "humidity": 68,
        "wind_speed": 9
    }
}
Display:

City
Temperature
Humidity
Wind speed
You will need to access values inside the nested "current" dictionary.

Example output:

City: Jacksonville
Temperature: 82
Humidity: 68
Wind Speed: 9

```python
weather = {
    "city": "Jacksonville",
    "current": {
        "temperature": 82,
        "humidity": 68,
        "wind_speed": 9
    }
}

print(f"City: {weather['city']}")
print(f"Temperature: {weather['current']['temperature']}")
print(f"Humidity: {weather['current']['humidity']}")
print(f"Wind Speed: {weather['current']['wind_speed']}")
```

Exercise 3: Work With a List Inside a Dictionary
Create:

forecast = {
    "city": "Jacksonville",
    "temperatures": [82, 84, 86, 83, 79]
}
Display:

The complete temperature list
The first temperature
The last temperature
The highest temperature
The lowest temperature
Use Python operations rather than manually typing the answers.



```python
forecast = {
    "city": "Jacksonville",
    "temperatures": [82, 84, 86, 83, 79]
}

print(f"Complete temperature list: {forecast['temperatures']}")
print(f"First temperature: {forecast['temperatures'][0]}")
print(f"Last temperature: {forecast['temperatures'][-1]}")
print(f"Highest temperature: {max(forecast['temperatures'])}")
print(f"Lowest temperature: {min(forecast['temperatures'])}")
```

Exercise 4: Process Multiple Records
Create:

data = {
    "cities": [
        {"name": "Jacksonville", "temperature": 82},
        {"name": "Orlando", "temperature": 87},
        {"name": "Tallahassee", "temperature": 84}
    ]
}
Use a loop to display each city's name and temperature.

Example:

Jacksonville: 82
Orlando: 87
Tallahassee: 84

```python
data = {
    "cities": [
        {"name": "Jacksonville", "temperature": 82},
        {"name": "Orlando", "temperature": 87},
        {"name": "Tallahassee", "temperature": 84}
    ]
}

for city in data["cities"]:
    print(f"{city['name']}: {city['temperature']}")
```

Exercise 5: Convert JSON Text to Python Data
Import Python's built-in json module:

import json
Create:

json_text = '{"name": "Alex", "age": 22, "major": "Computer Science"}'
Use:

json.loads()
to convert the JSON text into a Python dictionary.

Display:

Name
Age
Major
Then use type() to display the type of the converted object.

```python
import json

json_text = '{"name": "Alex", "age": 22, "major": "Computer Science"}'

json_data = json.loads(json_text)

print(f"Name: {json_data['name']}")
print(f"Age: {json_data['age']}")
print(f"Major: {json_data['major']}")
print(f"Type of converted object: {type(json_data)}")
```

Exercise 6: Convert Python Data to JSON
Create:

course = {
    "course": "COP 1000",
    "language": "Python",
    "active": True
}
Use:

json.dumps()
to convert the dictionary into JSON text.

Display the result.

Then display its data type using type().

```python
course = {
    "course": "COP 1000",
    "language": "Python",
    "active": True
}

json_text = json.dumps(course)

print(json_text)
print(f"Type of converted object: {type(json_text)}")
```
