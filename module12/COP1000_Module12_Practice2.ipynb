Exercise 2: Make Your First API Request
Use the Open-Meteo Geocoding API to search for Jacksonville.

Create:

url = "https://geocoding-api.open-meteo.com/v1/search"
Create a dictionary of parameters:

params = {
    "name": "Jacksonville",
    "count": 1,
    "language": "en",
    "format": "json"
}
Send the request:

response = requests.get(url, params=params)
Display:

response.status_code
A successful HTTP request normally returns:

200

```python
url = "https://geocoding-api.open-meteo.com/v1/search"

params = {
    "name": "Jacksonville",
    "count": 1,
    "language": "en",
    "format": "json"
}

response = requests.get(url, params=params)

print(response.status_code)
```

Exercise 5: Make a Weather Request
Store the latitude and longitude returned by the geocoding request.

Then use:

weather_url = "https://api.open-meteo.com/v1/forecast"
Create parameters similar to:

weather_params = {
    "latitude": latitude,
    "longitude": longitude,
    "current": "temperature_2m,wind_speed_10m",
    "temperature_unit": "fahrenheit",
    "wind_speed_unit": "mph"
}
Send the request using requests.get().

Convert the response using .json().

Display the complete response.

```python
latitude = location['latitude']
longitude = location['longitude']

weather_url = "https://api.open-meteo.com/v1/forecast"

weather_params = {
    "latitude": latitude,
    "longitude": longitude,
    "current": "temperature_2m,wind_speed_10m",
    "temperature_unit": "fahrenheit",
    "wind_speed_unit": "mph"
}

response = requests.get(weather_url, params=weather_params)

weather_data = response.json()

print(weather_data)
```

Exercise 6: Extract Current Weather
Access the "current" portion of the weather response.

Display:

Current temperature

Example output:

Wind Speed: 8.7 mph
Your values will depend on the weather at the time you run the program.

```python
temperature = weather_data['current']['temperature_2m']
wind_speed = weather_data['current']['wind_speed_10m']

print(f"Current Weather\nTemperature: {temperature} °F\nWind Speed: {wind_speed} mph")
```
